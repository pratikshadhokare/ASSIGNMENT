5.What is mutability in Python?

Mutability in Python refers to an object's ability to be altered after it has been created.

**Mutable objects:** These are objects that allow changes to their state or content after creation. You can modify, add, or remove elements from mutable objects.Eg-Lists,Dictionaries,Sets.

**Immutable objects**: These are objects whose state or content cannot be changed once they are created. Once an immutable object is created, it remains unchanged.
Eg-Tuple,Strings,Numbers.

6.Why are lists mutable, but tuples are immutable?

Lists are mutable in Python because they are designed for dynamic use cases where elements can be modified, added, or removed, making them suitable for handling changing data.

Tuples, on the other hand, are immutable to ensure data integrity, offering advantages like memory efficiency, faster performance, and hashability. Their immutability makes them ideal for storing fixed data that should not change, such as coordinates or constant values.

8.What are logical operators in Python?

9.What is type casting in Python?

10.What is the difference between implicit and explicit type casting?

**Implicit type casting** in Python happens automatically when Python converts data types during operations, like turning an integer into a float.

**Explicit type casting** is done manually by the programmer using functions like int(), float(), or str() to convert data types as needed.

11.What is the purpose of conditional statements in Python?

Conditional statements in Python are used to make decisions within a program by running specific blocks of code only when certain conditions are true. They use if, elif, and else to control the program's flow, allowing it to respond differently depending on various inputs or scenarios.


12.How does the elif statement work?

The elif statement in Python is used to test several conditions after the initial if statement. If the if condition is false, Python checks each elif condition in sequence, and once it finds one that is true, it executes that block of code and ignores the rest. If none of the conditions are true, the else block (if included) will run. This allows a program to choose between multiple possible actions.

13.What is the difference between for and while loops?

The difference between for and while loops in Python lies in how they handle repetition.

**A for loop** is used when the number of iterations is known or when iterating over a sequence, such as a list or range. It automatically processes each item in the sequence.

**A while loop** continues running as long as a condition is true, making it ideal for situations where the number of iterations depends on a dynamic condition. In summary, for loops are suited for fixed iterations, while while loops are used when the condition for repetition is variable.


14.Describe a scenario where a while loop is more suitable than a for loop?

A while loop is ideal when the number of iterations is unknown and depends on a changing condition. For instance, imagine monitoring the temperature of a machine. You could use a while loop to repeatedly check the temperature until it reaches a safe threshold. The loop would continue as long as the temperature is below the threshold and stop once it exceeds it. This makes a while loop the most suitable choice since the number of checks is not predetermined.

```python
#Write a Python program to convert user input from string to integer, float, and boolean types

user_input = input("Enter a value: ")

int_val = int(user_input) if user_input.isdigit() else None
float_val = float(user_input) if '.' in user_input and user_input.replace('.', '', 1).isdigit() else None
bool_val = user_input.lower() in ['true', '1', 'yes']

#convert values
print("Integer:", int_val)
print("Float:", float_val)
print("Boolean:", bool_val)
```

```python
#Write code to demonstrate type casting with list elements.

mixed_list = ['20', '7.5', 'True', '5000']

# Typecasting
int_list = [int(x) if x.isdigit() else None for x in mixed_list]
float_list = [float(x) if x.replace('.', '', 1).isdigit() else None for x in mixed_list]
bool_list = [x.lower() == 'true' if x.lower() in ['true', 'false'] else None for x in mixed_list]

# Display
print("Integer list:", int_list)
print("Float list:", float_list)
print("Boolean list:", bool_list)
```

```python
#Write a Python program to find the sum of all even numbers between 1 and 50.
even_sum = 0
for num in range(1, 51):
    if num % 2 == 0:  #even or not
        even_sum += num
print("Sum of all even numbers between 1 and 50:", even_sum)
```

```python
# Write a program to reverse a string using a while loop

input_string = input("Enter a string: ")
reversed_string = ""
index = len(input_string) - 1

while index >= 0:
    reversed_string += input_string[index]
    index -= 1

```

```python
#Write a Python program to calculate the factorial of a number provided by the user using a while loop.
num = int(input("Enter a number: "))
factorial = 1
i = 1
while i <= num:
    factorial *= i
    i += 1
print(f"The factorial of {num} is: {factorial}")
```
